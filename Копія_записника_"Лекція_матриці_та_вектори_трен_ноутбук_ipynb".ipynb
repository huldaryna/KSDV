```python
a = np.array([1, 2, 2])
b = np.array([2, 1, 2])

# TODO: обчисліть cos кута
cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))


# TODO: знайдіть кут у радіанах
angle_rad = np.arccos(cos_angle)


# TODO: переведіть кут у градуси
angle_deg = np.degrees(angle_rad)


print("Кут між векторами =", angle_deg)
```

## Завдання 13. Метод Крамера: система 2×2

Розв’яжіть систему методом Крамера:
\[
\begin{cases}
2x+y=5,\\
3x-2y=4.
\end{cases}
\]

Потрібно обчислити головний визначник та допоміжні визначники.

```python
A = np.array([[2,  1],
              [3, -2]], dtype=float)

b = np.array([5, 4], dtype=float)

# TODO: головний визначник
D = np.linalg.det(A)

# TODO: сформуйте матрицю Ax
A_x = A.copy()
A_x[:, 0] = b


# TODO: сформуйте матрицю Ay
A_y = A.copy()
A_y[:, 1] = b


# TODO: допоміжні визначники
D_x = np.linalg.det(A_x)
D_y = np.linalg.det(A_y)


# TODO: знайдіть x та y
x = D_x / D
y = D_y / D

print("D =", D)
print("x =", x)
print("y =", y)
```

## Завдання 14. Метод Крамера: система 3×3

Розв’яжіть систему методом Крамера:
\[
\begin{cases}
x+y+z=6,\\
2x-y+3z=9,\\
3x+2y-z=4.
\end{cases}
\]

Обчисліть \(D,D_x,D_y,D_z\) і знайдіть \(x,y,z\).

```python
A = np.array([[1,  1,  1],
              [2, -1,  3],
              [3,  2, -1]], dtype=float)

b = np.array([6, 9, 4], dtype=float)

# TODO: обчисліть D
D = np.linalg.det(A)

# TODO: створіть Ax, Ay, Az
A_x = A.copy()
A_x[:, 0] = b

A_y = A.copy()
A_y[:, 1] = b

A_z = A.copy()
A_z[:, 2] = b


# TODO: знайдіть Dx, Dy, Dz
D_x = np.linalg.det(A_x)
D_y = np.linalg.det(A_y)
D_z = np.linalg.det(A_z)


# TODO: знайдіть x, y, z
x = D_x / D
y = D_y / D
z = D_z / D


print("x =", x)
print("y =", y)
print("z =", z)
```

## Завдання 18. Метод Гауса: прямий хід

Для системи
\[
\begin{cases}
x+y+z=6,\\
2x+3y+z=10,\\
3x+y+2z=11
\end{cases}
\]
створіть розширену матрицю та виконайте **прямий хід методу Гауса**.

Потрібно отримати верхню трикутну матрицю.

```python
A = np.array([[1, 1, 1],
              [2, 3, 1],
              [3, 1, 2]], dtype=float)

b = np.array([6, 10, 11], dtype=float)

aug = np.column_stack((A, b))

# TODO: зануліть елемент під першим ведучим елементом у рядку 2
factor = aug[1, 0] / aug[0, 0]
aug[1] = aug[1] - factor * aug[0]


# TODO: зануліть елемент під першим ведучим елементом у рядку 3
factor = aug[2, 0] / aug[0, 0]
aug[2] = aug[2] - factor * aug[0]


# TODO: зануліть елемент під другим ведучим елементом у рядку 3
factor = aug[2, 1] / aug[1, 1]
aug[2] = aug[2] - factor * aug[1]


print("Матриця після прямого ходу:")
print(aug)
```

## Завдання 19. Метод Гауса: зворотний хід

Використайте верхню трикутну розширену матрицю
\[
\begin{pmatrix}
1&1&1&|&6\\
0&1&-1&|&-2\\
0&0&3&|&9
\end{pmatrix}
\]
та виконайте **зворотний хід методу Гауса**.

Знайдіть \(x,y,z\).

```python
aug = np.array([[1, 1,  1,  6],
                [0, 1, -1, -2],
                [0, 0,  3,  9]], dtype=float)

# TODO: знайдіть z
z = aug[2, 3] / aug[2, 2]


# TODO: знайдіть y
y = (aug[1, 3] - aug[1, 2] * z) / aug[1, 1]


# TODO: знайдіть x
x = (aug[0, 3] - aug[0, 1] * y - aug[0, 2] * z) / aug[0, 0]


print("x =", x)
print("y =", y)
print("z =", z)
```

## Завдання 20. Повний алгоритм методу Гауса

Реалізуйте розв’язування системи методом Гауса для
\[
\begin{cases}
2x+y-z=8,\\
-3x-y+2z=-11,\\
-2x+y+2z=-3.
\end{cases}
\]

Виконайте:
1. створення розширеної матриці;
2. прямий хід;
3. зворотний хід;
4. перевірку отриманого розв’язку.

```python
A = np.array([[ 2,  1, -1],
              [-3, -1,  2],
              [-2,  1,  2]], dtype=float)

b = np.array([8, -11, -3], dtype=float)

# TODO: створіть розширену матрицю
aug = np.column_stack((A, b))
n = len(b)

# Прямий хід
for i in range(n):
    # TODO: за потреби реалізуйте перестановку рядків
    if abs(aug[i, i]) < 1e-9:
        for k in range(i + 1, n):
            if abs(aug[k, i]) > abs(aug[i, i]):
                aug[[i, k]] = aug[[k, i]]
                break

    for j in range(i + 1, n):
        # TODO: обчисліть множник
        factor = aug[j, i] / aug[i, i]
        # TODO: виконайте елементарне перетворення рядка
        aug[j] = aug[j] - factor * aug[i]

# Зворотний хід
x = np.zeros(n)
for i in range(n - 1, -1, -1):
    # TODO: обчисліть суму вже відомих доданків
    known_sum = np.sum(aug[i, i+1:n] * x[i+1:n])
    # TODO: знайдіть x[i]
    x[i] = (aug[i, n] - known_sum) / aug[i, i]

# TODO: перевірте результат A @ x
check = A @ x

print("Розв'язок =", x)
print("Перевірка =", check)
```
